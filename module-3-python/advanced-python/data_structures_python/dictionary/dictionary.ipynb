```python
# what is dictionay in python 
# dictinary is an inbuilt data structures inside of python 
# dictionary called inside of {} with key:values paired 
# dict() function used dictinary call in key:values paired 
# dictionary key that is immutable will not changes and values is mutable can be changes 
# dictionary is mutables can be add | pop | del | update data in dictionary 
# dictionay can stored dublicate values but can not stored dublicate key 

# how to access dictionary 

# employee={1:"brijesh",2:"brijesh", 3:"het",4:"devyang"}
# print(employee)

# iterate a  dictinary key 
# employee={1:"brijesh",2:"brijesh", 3:"het",4:"devyang"}
# #print(employee)
# for key in employee:
#     print(key)


#iterate a dictionary values 

employee={1:"krish",2:"brijesh", 3:"het",4:"devyang"}
#print(employee)
for value in employee.values():
    print(value)
```

```python
# accessing a dictionary values via its key 
# dict() used as a function or constructor of dictionary
emp=dict({"name":"brijesh","age":35,"salary":89600})
print(emp)
print(emp["name"])
print(emp["age"])
print(emp["salary"])
```

```python
# get a dictionary values used get() method
emp=dict({"name":"brijesh","age":35,"salary":89600})
print(emp.get("name"))
print(emp.get("age"))
print(emp.get("salary"))
```

```python
# nested dictionary 
fruits={1:"banana",2:"apple",3:{"a":"pineapple","b":"guvava"}}
print(fruits)
print(fruits[2])
print(fruits[3])
print(fruits[3]["b"])
```

```python
# dictionary is mutable can be added | del | update 
employee={1:"krish",2:"brijesh", 3:"het",4:"devyang"}
# add a new items in dictionary 
employee[5]="rajesh"
employee[6]="lokesh"
print(employee)
```

```python
#update a dictionary 
employee={1:"krish",2:"brijesh", 3:"het",4:"devyang"}
employee[4]="rohit"
print(employee)
```

```python
#del a dictionary 
employee={1:"krish",2:"brijesh", 3:"het",4:"devyang"}
del employee[2]
print(employee)
```

```python
#delete a last items from dictionary 
employee={1:"krish",2:"brijesh", 3:"het",4:"devyang"}
employee.pop(1)
print(employee)
```

```python
# iterate a dictionary via key and values 
# iterate by key 
# employee={1:"krish",2:"brijesh", 3:"het",4:"devyang"}
# for key in employee:
#     print(key)


# iterate by values 
# employee={1:"krish",2:"brijesh", 3:"het",4:"devyang"}
# for value in employee.values():
#     print(value)

# iterate via key and values 
employee={1:"krish",2:"brijesh", 3:"het",4:"devyang"}
for key,value in employee.items():
    print(f"{key}:{value}")
```

```python
# string formatter print multiple sring using f"{}"
employee={1:"krish",2:"brijesh", 3:"het",4:"devyang"}
for key,value in employee.items():
    print(f"{key}:{value}")
```

```python
# clear a dictionary or empty a dictionary 
employee={1:"krish",2:"brijesh", 3:"het",4:"devyang"}
employee.clear()
print(employee)
```

```python
# popitem() it is usd to delete last items 
employee={1:"krish",2:"brijesh", 3:"het",4:"devyang"}
employee.popitem()
print(employee)
```

```python
employee={1:"krish",2:"brijesh", 3:"het",4:"devyang"}
employee.update({4:"kumar"})
print(employee)
```

```python
# can not stored duplicate key but can be stored dublicate values 
# employee={1:"krish",1:"brijesh"}
# print(employee)

employee={1:"brijesh",2:"brijesh"}
print(employee)
```

```python
# merge a dictionary 
dict1={1:"brijesh"}
dict2={2:"rajesh"}
dict1.update(dict2)
print(dict1)
```

```python
# sum of values in dictionary
marks={"math":50,"chem":80,"comp":95}
print(sum(marks.values()))
```
